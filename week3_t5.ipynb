```python
print(df.head())
print(df.info())
```

```python
df['Shipdate'] = pd.to_datetime(df['Shipdate'])
df['Year'] = df['Shipdate'].dt.year
df['Month'] = df['Shipdate'].dt.month_name()
df['Amount_per_Box'] = df['Amount'] / df['Boxes']

# Load
df.to_excel("processed_chocolate_sales.xlsx", index=False)

print("ETL Pipeline Completed Successfully!")
print(df.head())
```
