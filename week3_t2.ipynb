```python
import pandas as pd

df = pd.read_excel("cleaned_titanic_dataset.xlsx")

print(df.head())
```

```python
print(df.info())
print(df.describe())
```

```python
df['gender'] = df['gender'].map({
    'male':0,
    'female':1
})
```

```python
df['embarked'] = df['embarked'].map({
    'S':0,
    'C':1,
    'Q':2
})
```

```python
df['family_type'] = df['family'].apply(
    lambda x: 0 if x==0 else 1
)
```

```python
df['fare_category'] = pd.cut(
    df['fare'],
    bins=3,
    labels=[0,1,2]
)
```

```python
df['age_group'] = pd.cut(
    df['age'],
    bins=[0,12,19,60,100],
    labels=[0,1,2,3]
)
```

```python
df = df.drop(
    ['sn','name','date'],
    axis=1
)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[['age','fare']] = scaler.fit_transform(
    df[['age','fare']]
)
```
